### Question 11:

Using zero-padding in convolutional layers can preserve the input
spatial dimensions when using a $3 \times 3$ kernel/filter, stride 1,
and padding $P = 1$. True or False?

We have the formula 

$$
Output size = \frac{N+2P-K}{S}+1
$$

For $K=3, S=1, P=1$

$$
\frac{N+2*1-3}{1}+1 = \frac{N-1}{1}+1 = N
$$

In other words we have the same input and output size with these parameters

### Question 16:

What is Backpropagation Through Time (BPTT) and why is it required for training RNNs?

BPTT is an extentions of normal backpropogation used to train RNN. It works by treating the RNN as a deep NN with shared weights.

We need this extension since our hidden state depends recursively on earlier states of the RNN. 

### Question 17:

What does a sliding window do? And why would we use it?

A sliding window extracts a fixed length, consecutive segments from a sequence by moving a window step by step across the data.

We use it to produce training damples from a time series, detecy local patterns, and enable operations like convolutions.
